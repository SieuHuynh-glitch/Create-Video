### Bước 5: Đóng gói và Tải toàn bộ 12 Clip về máy tính
Sau khi Colab sinh xong, đoạn code dưới đây sẽ tự động nén toàn bộ 12 file `shot_01.mp4` -> `shot_12.mp4` thành file `natural-cycle-v2-clips.zip` và kích hoạt tải về máy của bạn.

````python
# 5. Đóng gói ZIP và tải về máy tính
import zipfile
from google.colab import files
from pathlib import Path

media_dir = Path("runs/natural-cycle-v2/media")
zip_path = Path("natural-cycle-v2-clips.zip")

mp4_files = sorted(list(media_dir.glob("shot_*.mp4")))
print(f"📦 Tìm thấy {len(mp4_files)} clip đã sinh thành công:")
with zipfile.ZipFile(zip_path, "w", zipfile.ZIP_DEFLATED) as zipf:
    for f in mp4_files:
        print(f"  + Đóng gói: {f.name}")
        zipf.write(f, arcname=f.name)

print(f"\n🎉 Đã nén thành công: {zip_path.name} ({zip_path.stat().st_size / 1024 / 1024:.2f} MB)")
print("📥 Đang tải file zip về máy tính của bạn...")
files.download(str(zip_path))
````
